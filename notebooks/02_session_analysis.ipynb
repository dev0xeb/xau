# 02 - XAUUSD Session Dynamics & Return Distribution Analysis

Analyzes empirical price movement, range, and return distributions across Asian, London, NY, and Overlap sessions.

```python
import pandas as pd

df = pd.read_parquet('../data/processed/features/XAUUSD_M1_features.parquet')
session_stats = df.groupby('session_label')[['high_low_range', 'ret_abs', 'estimated_spread_usd']].mean()
print(session_stats)
```
